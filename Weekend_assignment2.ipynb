## Q1. What's the difference between a Series and a DataFrame?

A Series is a one-dimensional data structure that stores a single column of data with an index. A DataFrame is a two-dimensional table consisting of rows and multiple columns, where each column is a Series.

An operation like `.head()` works on both a Series and a DataFrame. Operations such as selecting multiple columns (`df[['col1','col2']]`) are specific to a DataFrame.

## Q2. What does inplace=True do?

`inplace=True` modifies the original DataFrame instead of creating a new one. This can save memory, but it often makes code harder to read and debug.

The pandas team discourages its use because many inplace operations still create temporary copies internally. I would avoid using it when cleaning data so that the original dataset remains available if something goes wrong.

## Q3. Difference between fillna() and dropna()

`fillna()` replaces missing values with a specified value, while `dropna()` removes rows or columns containing missing values.

For example, if customer ages are missing, I would use `fillna()` with the average age. If only a few records have missing IDs that cannot be recovered, I would use `dropna()` to remove those rows.

## Q4. What does pd.to_numeric(errors='coerce') do?

`pd.to_numeric(errors='coerce')` converts valid values into numbers and changes invalid values into `NaN` instead of raising an error.

Without `errors='coerce'`, pandas stops execution whenever it encounters an invalid value. For messy real-world datasets, I would always use `errors='coerce'` because it safely handles incorrect entries.

## Q5. Difference between .loc and .iloc

`.loc` selects rows and columns using labels, while `.iloc` selects them using integer positions.

Example:

- `df.loc[2]` returns the row with label 2.
- `df.iloc[2]` returns the third row.

If the DataFrame index is changed, such as `[10,11,12]`, then `df.loc[10]` returns the first row, whereas `df.iloc[0]` also returns the first row. In this case, `df.loc[0]` would not exist.

## Q6. Difference between map(), apply(), and replace()

`map()` is mainly used to transform Series values using a dictionary or function.

`apply()` applies a custom function to each value and is useful for more complex transformations.

`replace()` substitutes specific existing values with new ones without applying a function.

I would use `map()` for category mapping, `apply()` for calculations, and `replace()` for correcting inconsistent values.

## Q8. Difference between pd.cut() and pd.qcut()

`pd.cut()` divides data into fixed numeric ranges, while `pd.qcut()` divides data so that each group contains nearly the same number of observations.

If I want to classify annual income into Low, Mid, and High using fixed salary ranges, I would use `pd.cut()`. If I want each category to contain an equal number of customers, I would use `pd.qcut()`.

````python
import pandas as pd
import numpy as np
````

````python
df = pd.read_csv("indian_startups_fundingDADS.csv")

# Display first few rows
df.head()
````

````python
# -----------------------------
# Rename columns to snake_case
# -----------------------------
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
)

# -----------------------------
# Clean text columns
# -----------------------------
df["startup_name"] = df["startup_name"].str.strip().str.title()
df["city"] = df["city"].str.strip().str.title()

# -----------------------------
# Standardize city names
# -----------------------------
city_map = {
    "Bangalore": "Bengaluru",
    "Blr": "Bengaluru",
    "Bombay": "Mumbai",
    "Gurgaon": "Gurugram",
    "Delhi": "New Delhi",
    "Calcutta": "Kolkata",
    "Madras": "Chennai"
}

df["city"] = df["city"].replace(city_map)

# -----------------------------
# Standardize Industry
# -----------------------------
industry_map = {
    "Fin Tech":"Fintech",
    "Fintech":"Fintech",
    "Ed Tech":"Edtech",
    "Edtech":"Edtech",
    "Health Tech":"HealthTech",
    "Healthtech":"HealthTech",
    "Health Tech ":"HealthTech",
    "E-Commerce":"Ecommerce",
    "Ecommerce":"Ecommerce",
    "Saas":"SaaS",
    "Ai":"AI"
}

df["industry"] = (
    df["industry"]
    .str.strip()
    .str.title()
    .replace(industry_map)
)

# -----------------------------
# Standardize Funding Stage
# -----------------------------
stage_map = {
    "Pre Seed":"Pre-Seed",
    "Pre-seed":"Pre-Seed",
    "Seed":"Seed",
    "Series A":"Series A",
    "Series B":"Series B",
    "Series C":"Series C",
    "Series D":"Series D",
    "Series E":"Series E",
    "Bridge":"Bridge"
}

df["funding_stage"] = (
    df["funding_stage"]
    .str.strip()
    .str.title()
    .replace(stage_map)
)

# -----------------------------
# Clean Amount
# -----------------------------
df["amount"] = (
    df["amount"]
    .astype(str)
    .str.replace("₹","", regex=False)
    .str.replace("INR","", regex=False)
    .str.replace("Cr","", regex=False)
    .str.replace("Crore","", regex=False)
    .str.replace(",","", regex=False)
    .str.replace("Undisclosed","", regex=False)
    .str.strip()
)

df["amount"] = pd.to_numeric(df["amount"], errors="coerce")

# -----------------------------
# Convert Date
# -----------------------------
df["funding_date"] = pd.to_datetime(
    df["funding_date"],
    errors="coerce",
    dayfirst=True
)

# -----------------------------
# Remove duplicates
# -----------------------------
df = df.drop_duplicates()

df.head()
````

````python
# Create funding_efficiency column

df["funding_efficiency"] = df["amount"] / (2024 - df["founded_year"])

top10 = df.sort_values(
    by="funding_efficiency",
    ascending=False
)[
    [
        "startup_name",
        "industry",
        "amount",
        "founded_year",
        "funding_efficiency"
    ]
].head(10)

print(top10)
````

````python
# Funding Tier

df["funding_tier"] = df["amount"].apply(
    lambda x: "Unicorn" if pd.notna(x) and x >= 1000
    else "Mega" if pd.notna(x) and x >= 500
    else "Large" if pd.notna(x) and x >= 100
    else "Mid" if pd.notna(x) and x >= 10
    else "Small / Unknown"
)

print(df["funding_tier"].value_counts())
````

````python
# Founded Year Eras

df["era"] = pd.cut(
    df["founded_year"],
    bins=[0, 2009, 2017, float("inf")],
    labels=["Early", "Growth", "Recent"]
)

print(df["era"].value_counts())
````

````python
# Investor Count

df["investor_count"] = df["investors"].apply(
    lambda x: 0 if pd.isna(x) else len(str(x).split(", "))
)

print(df["investor_count"].value_counts())
````
